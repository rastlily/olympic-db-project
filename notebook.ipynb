### 1. ERD Design  ([`ERD_olimpics.sql`](../sql/ERD_olimpics.sql))  

The first step involved creating a normalized logical schema for the Olympic database.  
Tables created:

- **Athlet** – Athlete information (ID, Name, Gender, Age, Height, Weight, NOC)  
- **Land** – Country data (NOC, Name)  
- **Sportart** – Sport disciplines  
- **Event** – Olympic events linked to sport and year  
- **Medaille** – Medal types  
- **Teilnahme** – Participation linking athletes, events, medals, and countries  

Relationships were implemented using foreign keys to ensure referential integrity.

### 2. Database Creation ([`CREATE_DATABASE_olimpics.sql`](../sql/CREATE_DATABASE_olimpics.sql))
The ERD schema was implemented in MySQL under the database `olympics`.  
Simplified structure for working with CSV data:

| Table | Description |
|-------|-------------|
| `athletes` | Athlete ID, Name, Sex |
| `events` | Sport and event names |
| `countries` | Country codes and names |
| `athlete_events` | Participation records (year, medal, age, height, weight, NOC) |

```sql
LOAD DATA INFILE 'C:\\ProgramData\\MySQL\\MySQL Server 9.1\\Uploads\\athletes.csv'
INTO TABLE athletes
FIELDS TERMINATED BY ';'
ENCLOSED BY '"'
LINES TERMINATED BY '\n'
IGNORE 1 ROWS;  



- Additional handling of missing values during import:

```sql
SET 
    Medal = NULLIF(@Medal, ''),
    Age = NULLIF(@Age, ''),
    Height = NULLIF(@Height, ''),
    Weight = NULLIF(@Weight, '');

- After import, the database was fully populated and ready for analysis.

### 4. Query Development: [`QUERIES_olympics.sql`](./sql/QUERIES_olympics.sql)

**Intermediate Queries:**  


- Which countries won the most medals overall?  

Question: Which countries have the highest total medal count across all Olympic Games?  
Insight: Shows top-performing nations by total medal count — reveals historical dominance patterns.


- Average age, height, and weight of medalists by sport  

Question: What are the average physical characteristics of medal-winning athletes for each sport?  
Insight: Highlights demographic profiles, showing which sports favor younger or taller athletes.


- Gender distribution of participants per year  

Question: How did the ratio of male and female athletes change over time?  
Insight: Shows the evolution of gender representation at the Olympics, useful for diversity analysis.  


**Advanced Queries:**  


- Medal conversion rate by country  

Question: What percentage of a country’s total athlete participations resulted in medals?  
Insight: Identifies countries most efficient at turning participation into medals.

- Most successful athletes (multi-medal winners)  

Question: Who are the top athletes by total number of medals won?  
Insight: Highlights all-time greatest athletes by medal count.

- Medal trends over decades  

Question: How has medal distribution evolved across decades?  
Insight: Visualizes long-term dominance patterns and historical shifts.

- Sports with the highest medal diversity  

Question: Which sports have the widest distribution of medal-winning countries?  
Insight: Identifies sports with globally balanced competition.

- Yearly country rankings by total medals  

Question: How did each country rank by total medals in each Olympic year?  
Insight: Shows country rankings over time — highlights rising nations and long-term dominance.

- Cumulative medal count trend per country  

Question: How has each country's total medal count accumulated over time?  
Insight: Displays cumulative performance — useful for analyzing historical growth and trajectories.  

*simple queries (basic stats, filters) are available via an interactive Streamlit dashboard for exploration and visualization.





### 5. Visualization ([`streamlit_app.py`](streamlit_app.py))  

An interactive Streamlit dashboard was created to explore Olympic data with dynamic filters. Key features include:

- Filterable Data Table: Users can filter by country, year, and medal type to display detailed athlete participation and performance records.  

- Medal Distribution: Bar chart showing the count of Gold, Silver, and Bronze medals for the selected country and year.  

- Age Distribution by Medal: Box plot visualizing the age distribution of athletes based on medal type.  

- Top 10 Countries by Athlete Count: Bar chart highlighting the countries with the most athletes in the selected year.  

- Medal Trends Over Time: Line chart displaying the number of medals earned by the selected country across multiple Olympic years.  

- Preview of Raw Tables: Optional checkbox to view the first rows of the source CSVs (`athletes`, `events`, `countries`, `athlete_event`) and the merged dataset.

All charts use Plotly for interactive visualization and are dynamically updated based on user-selected filters.  
The dashboard provides an intuitive interface for exploring athlete performance, medal counts, and historical trends.  

![](../olympics_streamlit_dashboard/demo.GIF)


### 6. Summary of Project Structure

| Stage | File | Purpose |
|-------|------|---------|
| ERD Design | [`ERD_olympics.sql`](sql/ERD_olympics.sql) | Logical database design and relationships |
| Database Creation | [`CREATE_DATABASE_olympics.sql`](sql/CREATE_DATABASE_olympics.sql) | Physical schema implementation in MySQL |
| Data Loading | [`LOAD_DATA_to_db_olympics.sql`](sql/LOAD_DATA_to_db_olympics.sql) | CSV import with data preprocessing |
| Query Development | [`QUERIES_olympics.sql`](sql/QUERIES_olympics.sql) | SQL analytics |
| Visualization | [`app.py`](streamlit_app.py) | Interactive dashboard for basic queries |
